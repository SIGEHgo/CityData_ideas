```python
with conn.cursor() as cur:
    # 1. Activa la extensión PostGIS
    cur.execute("CREATE EXTENSION IF NOT EXISTS postgis;")

    # 2. Añade una nueva columna de geometría
    cur.execute("""
        ALTER TABLE public.prueba
        ADD COLUMN IF NOT EXISTS geom_origen geometry(Point, 4326);
    """)

    # 3. Llena la columna con datos de latitud/longitud
    cur.execute("""
        UPDATE public.prueba
        SET geom_origen = ST_SetSRID(ST_MakePoint(overlap_origin_long, overlap_origin_lat), 4326)
        WHERE overlap_origin_lat IS NOT NULL AND overlap_origin_long IS NOT NULL;
    """)

    # Guarda los cambios
    conn.commit()
```
